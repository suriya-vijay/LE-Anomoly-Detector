# 02 - Features over time

**Plain language:** each 1-second slice becomes a short list of numbers. Here we plot the easy-to-understand ones over each whole recording:

- **Loudness (dB)**: jumps up when the motor runs, higher again while cutting.
- **Main pitch**: mostly motor speed; dips slightly when the blade is loaded.
- **Brightness**: dull hum (low) vs. scratchy/screechy (high).
- **Pulsing strength**: near zero for a steady sound, high for a "duh duh duh" thumping.

The bottom picture is the 64-band *tone fingerprint*, one column per slice. Compare the good cut with the weird one: which rows look different, and when?
